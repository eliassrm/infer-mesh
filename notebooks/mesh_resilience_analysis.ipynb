# INFER-MESH Resilience Analysis

Starter analysis for telemetry generated by controlled mesh degradation experiments. The sample dataset illustrates the initial CSV contract; production analysis should include experiment metadata and repeated matched runs.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

candidate_paths = [Path('../data/sample_experiment.csv'), Path('data/sample_experiment.csv')]
data_path = next(path for path in candidate_paths if path.exists())
df = pd.read_csv(data_path, parse_dates=['timestamp'])
df
```

## Event summary

Group observations by experimental event window to compare normal and degraded behavior.

```python
summary = (
    df.groupby('event_label')
      .agg(
          samples=('timestamp', 'count'),
          median_latency_ms=('latency_ms', 'median'),
          median_loss=('packet_loss', 'median'),
          median_tq=('batman_tq', 'median'),
          median_throughput_mbps=('throughput_mbps', 'median'),
      )
)
summary
```

## Telemetry timeline

```python
fig, axes = plt.subplots(3, 1, figsize=(11, 8), sharex=True)

axes[0].plot(df['timestamp'], df['latency_ms'], marker='o', label='Latency (ms)')
axes[0].set_ylabel('ms')
axes[0].legend(loc='upper left')

axes[1].plot(df['timestamp'], df['throughput_mbps'], marker='o', color='tab:green', label='Throughput (Mbps)')
axes[1].set_ylabel('Mbps')
axes[1].legend(loc='upper left')

axes[2].plot(df['timestamp'], df['batman_tq'], marker='o', color='tab:orange', label='batman-adv TQ')
axes[2].set_ylabel('TQ')
axes[2].set_xlabel('Timestamp (UTC)')
axes[2].legend(loc='upper left')

fig.suptitle('INFER-MESH sample telemetry')
fig.autofmt_xdate()
fig.tight_layout()
plt.show()
```
